# F2 — Milan Original (Pointer)

F2 does not introduce a new Milan chapter. It reuses the **same UA Advanced Chapter 1 pattern** already retyped in F1's [01_milan_original.ipynb](../ua_advanced_ch01_setup/01_milan_original.ipynb), just applied to a different OSM feature class.

**The one substitution:**

| Where | Milan / F1 | F2 |
| :--- | :--- | :--- |
| OSM query tag | `tags={'building': True}` | `tags={'highway': ['motorway']}` |
| Geometry filter | `Polygon` / `MultiPolygon` | `LineString` / `MultiLineString` |
| What it returns | ~46K building footprints | ~245 IH-35 way segments |
| What it becomes | area context, land-use joins | corridor for buffer, candidates, spacing |

That is the only conceptual change. Every other step from F1 — the two-CRS discipline, the `admin_poly` input, the unique-key assertion, the `.gpkg` cache — carries over unchanged.

If you want to re-read the underlying pattern, open F1's Milan retype: [../ua_advanced_ch01_setup/01_milan_original.ipynb](../ua_advanced_ch01_setup/01_milan_original.ipynb).
